```python
from dataCuration import clean_psm
import os
from pathlib import Path
import pandas as pd
```

```python
libs = ["6mer","6mer2","8mer","8mer2","9mer","10mer","12mer","13mer","15mer","varLen2"]
method1_libs = [f'{lib}_{term}-term_{tag}.tsv' for term in ["R","K"] for tag in ["D-Phe","L-Phe","LF","2plex"] for lib in ["6mer","9mer"]]
method2_libs = [f'12mer_{term}-term_{tag}.tsv' for term in ["R","K"] for tag in ["D-Phe","L-Phe","LF","2plex"]]+["15mer_K-term_LF.tsv","15mer_R-term_LF.tsv"]
for lib in libs:
    input_path = Path(f"/home/amirabbas-kazeminia/Projects/DeepRT/data/rawdata/{lib}/psm")
    input_files = [f for f in input_path.iterdir() if f.is_file()]
    output_path = Path(f"/home/amirabbas-kazeminia/Projects/DeepRT/data/curatedData/{lib}")
    output_path.mkdir(parents=True, exist_ok=True)
    for file in input_files:
        file_name = file.name
        output_path_dir = str(output_path)+'/'+file_name[:-4]+'.csv'
        method = "method_3"
        if file_name in method1_libs:
            method = "method_1"
        elif file_name in method2_libs:
            method = "method_2"
        try:
            clean_psm(str(file), output_path_dir, method)
        except:
            print(file_name)
```

# Data Merger

```python
libs = ["6mer","6mer2","8mer","8mer2","9mer","10mer","12mer","13mer","15mer","varLen2"]
data = []
for lib in libs:
    input_path = Path(f"/home/amirabbas-kazeminia/Projects/DeepRT/data/curatedData/{lib}")
    input_files = [f for f in input_path.iterdir() if f.is_file()]
    for file in input_files:
        df = pd.read_csv(file)
        data.append(df)
merged = pd.concat(data, ignore_index=True)
merged.to_csv("ML_DATA.csv",index=False)
```
